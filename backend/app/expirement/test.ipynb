# Agile Project Dataset 2024 - Data Exploration

This notebook loads and explores the Agile Project Dataset from Kaggle.

```python
# Import required libraries
import kagglehub
import pandas as pd
import os
import glob

print("✅ Libraries imported successfully")
```

```python
# Download the dataset from Kaggle
# This will download the entire dataset to your local machine
print("📥 Downloading dataset...")

# Download latest version
path = kagglehub.dataset_download("digrok/agile-project-dataset-2024")

print(f"✅ Dataset downloaded to: {path}")
print(f"\n📂 Files in dataset:")

# List all files in the downloaded dataset
all_files = []
for root, dirs, files in os.walk(path):
    for file in files:
        file_path = os.path.join(root, file)
        file_size = os.path.getsize(file_path) / (1024 * 1024)  # Convert to MB
        ext = os.path.splitext(file)[1]
        all_files.append((file, file_path, file_size, ext))
        print(f"  • {file} ({file_size:.2f} MB) [{ext}]")

print(f"\n📊 Total files: {len(all_files)}")
```

```python
# Load the data files
# This will automatically detect and load common data formats

df = None
loaded_file = None

print("🔍 Attempting to load data automatically...\n")

# Priority order: CSV, TSV, JSON, Excel, Parquet, Feather
for filename, filepath, size, ext in all_files:
    ext_lower = ext.lower()
    
    try:
        if ext_lower in ['.csv', '.txt'] and df is None:
            print(f"📊 Loading CSV: {filename}")
            df = pd.read_csv(filepath)
            loaded_file = filename
            break
        elif ext_lower == '.tsv' and df is None:
            print(f"📊 Loading TSV: {filename}")
            df = pd.read_csv(filepath, sep='\t')
            loaded_file = filename
            break
        elif ext_lower in ['.json', '.jsonl'] and df is None:
            print(f"📊 Loading JSON: {filename}")
            df = pd.read_json(filepath)
            loaded_file = filename
            break
        elif ext_lower in ['.xlsx', '.xls'] and df is None:
            print(f"📊 Loading Excel: {filename}")
            df = pd.read_excel(filepath)
            loaded_file = filename
            break
        elif ext_lower == '.parquet' and df is None:
            print(f"📊 Loading Parquet: {filename}")
            df = pd.read_parquet(filepath)
            loaded_file = filename
            break
        elif ext_lower == '.feather' and df is None:
            print(f"📊 Loading Feather: {filename}")
            df = pd.read_feather(filepath)
            loaded_file = filename
            break
    except Exception as e:
        print(f"⚠️  Failed to load {filename}: {str(e)}")
        continue

if df is not None:
    print(f"\n✅ Successfully loaded: {loaded_file}")
    print(f"   Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
    print(f"   Memory usage: {df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
else:
    print("\n❌ Could not load any data file automatically.")
    print("\n💡 Run the next cell for manual loading instructions.")
```

```python
# Manual Data Loading Instructions
# Use this cell if automatic loading didn't work

if 'path' in locals() and df is None:
    print("📂 Dataset location: " + path)
    print("\n🛠️  Manual Loading Examples:")
    print("="*60)
    
    print("\n1️⃣  For CSV files:")
    print("   df = pd.read_csv(os.path.join(path, 'filename.csv'))")
    
    print("\n2️⃣  For JSON files:")
    print("   df = pd.read_json(os.path.join(path, 'filename.json'))")
    print("   # or for JSON Lines:")
    print("   df = pd.read_json(os.path.join(path, 'file.jsonl'), lines=True)")
    
    print("\n3️⃣  For Excel files:")
    print("   df = pd.read_excel(os.path.join(path, 'filename.xlsx'))")
    
    print("\n4️⃣  For Parquet files:")
    print("   df = pd.read_parquet(os.path.join(path, 'filename.parquet'))")
    
    print("\n5️⃣  For compressed files:")
    print("   df = pd.read_csv(os.path.join(path, 'file.csv.gz'), compression='gzip')")
    print("   df = pd.read_csv(os.path.join(path, 'file.csv.zip'))")
    
    print("\n6️⃣  For multiple files (combine them):")
    print("   files = glob.glob(os.path.join(path, '*.csv'))")
    print("   dfs = [pd.read_csv(f) for f in files]")
    print("   df = pd.concat(dfs, ignore_index=True)")
    
    print("\n" + "="*60)
    print("📋 Available files in dataset:")
    for filename, filepath, size, ext in all_files:
        print(f"   • {filename} ({size:.2f} MB)")
    
    print("\n💡 TIP: Copy one of the examples above, replace 'filename'")
    print("   with your actual file name, and run it in a new cell.")
    
elif df is not None:
    print("✅ Data already loaded!")
    print(f"   Using: {loaded_file}")
else:
    print("⚠️  Please run the download cell first.")

# Example: Uncomment and modify this to manually load:
# df = pd.read_csv(os.path.join(path, 'your_actual_filename.csv'))
# print(f"✅ Manually loaded: {df.shape}")
```

```python
# Display basic information about the dataset
if 'df' in locals() and df is not None:
    print("📋 Dataset Overview")
    print("=" * 60)
    
    print("\n📊 First 5 records:")
    display(df.head())
    
    print("\n📈 Dataset Info:")
    df.info()
    
    print("\n📉 Statistical Summary:")
    display(df.describe())
    
    print("\n🔍 Column Names:")
    for i, col in enumerate(df.columns, 1):
        print(f"  {i}. {col}")
    
    print("\n❓ Missing Values:")
    missing = df.isnull().sum()
    if missing.sum() > 0:
        missing_pct = (missing / len(df)) * 100
        missing_df = pd.DataFrame({
            'Missing Count': missing[missing > 0],
            'Percentage': missing_pct[missing > 0].round(2)
        })
        display(missing_df)
    else:
        print("  ✅ No missing values found!")
else:
    print("⚠️  DataFrame not loaded. Please run the previous cells first.")
```

```python
# Quick data exploration
if 'df' in locals() and df is not None:
    print("🎯 Quick Analysis")
    print("=" * 60)
    
    # Show data types
    print("\n📊 Data Types:")
    print(df.dtypes)
    
    # Show unique value counts for categorical columns
    print("\n🏷️  Unique Values per Column:")
    for col in df.columns:
        unique_count = df[col].nunique()
        print(f"  • {col}: {unique_count:,} unique values")
    
    # Sample values for categorical columns
    print("\n🔤 Sample Values (for categorical columns):")
    for col in df.select_dtypes(include=['object']).columns[:5]:  # First 5 categorical
        print(f"\n  {col}:")
        print(f"    {df[col].value_counts().head(3).to_dict()}")
else:
    print("⚠️  DataFrame not loaded. Please run the previous cells first.")
```
